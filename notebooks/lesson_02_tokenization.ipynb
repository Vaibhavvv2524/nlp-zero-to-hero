# Lesson 02 — Tokenization

**Tokenization** means splitting text into smaller pieces that a computer can work with.

There are two useful kinds:

- **Sentence tokenization:** split a paragraph into sentences.
- **Word tokenization:** split a sentence into words or word pieces.

This lesson uses small English examples so the idea is easy to see. The final project applies the same idea to Hindi.

## Before you start

The next cell downloads two small NLTK data packages. You only need to do this once in an environment.


```python
import nltk

for resource in ["punkt", "punkt_tab"]:
    try:
        nltk.download(resource, quiet=True)
    except Exception as error:
        print("Could not download", resource, ":", error)

print("NLTK is ready.")
```

## Step 1 — Split a paragraph into sentences

Read the example, then run the cell. Notice that the full stop is used as a clue.


```python
from nltk.tokenize import sent_tokenize, word_tokenize

text = "I love NLP. It is fun!"

print("Sentences:", sent_tokenize(text))
print("Words:", word_tokenize(text))
```

**What you should see:** a list with two sentences, then a list of words and punctuation.

The punctuation marks are tokens too. That matters because punctuation is exactly what the project tries to restore.


## Step 2 — Try a Hindi sentence

Hindi words are usually separated by spaces, so a simple `split()` works for words. But the danda stays attached to the last word. That is the problem the project solves.


```python
hindi = "मुझे एनएलपी पसंद है। यह मज़ेदार है।"

print("Simple word split:", hindi.split())
```

**What you should see:** `है।` appears as one token. The model must learn that the danda means “end of sentence” and should be treated as punctuation, not part of the word.

## Why this matters

The final pipeline works with Hindi words. It must decide where punctuation belongs without changing the words themselves.
