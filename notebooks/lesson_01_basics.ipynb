## Step 1 — Run Python

The next cell prints the Python version. This is the first check that your environment works.


````python
import sys

print("Python version:", sys.version.split()[0])
print("You are running Python. Any cell can print text like this.")
````

**What you should see:** a Python version. The exact version is not important. What matters is that the cell runs without an error.


## Step 2 — Meet NumPy and pandas

Two tools appear in almost every data project:

- **NumPy** helps us work with lists of numbers.
- **pandas** helps us work with tables, like a spreadsheet.

The next cell checks that both are installed.


````python
import numpy as np
import pandas as pd

print("NumPy version:", np.__version__)
print("pandas version:", pd.__version__)
````

## Step 3 — A tiny NumPy example

NumPy makes simple maths on a list of numbers easy.


````python
numbers = np.array([10, 20, 30, 40])

print("Numbers:", numbers)
print("Average:", numbers.mean())
````

## Step 4 — A tiny pandas example

A pandas table has named columns. This table stores words and how often they appear.


````python
words = pd.DataFrame({
    "word": ["नमस्ते", "दुनिया", "NLP"],
    "count": [2, 1, 3],
})

print(words)
````

**What you should see:** a small table with two columns, `word` and `count`.

You do not need to memorise NumPy or pandas. The point is to see that Python can hold numbers and tables, and that your environment is ready.
