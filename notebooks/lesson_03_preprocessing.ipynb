# Lesson 03 — Preprocessing

**Preprocessing** means preparing text before a model uses it.

This lesson shows four common steps:

1. **Cleaning:** remove links, mentions, and extra symbols.
2. **Tokenization:** split text into words.
3. **Stopword removal:** remove very common words such as “the” or “is”.
4. **Stemming and lemmatization:** reduce words to a simpler base form.

These examples use English so the changes are easy to see. The final Hindi project uses its own preprocessing.

## Step 0 — Find the project folder

The next cell finds the repository root and downloads the small NLTK data packages used by the lesson.


```python
from pathlib import Path
import sys

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

import nltk

for resource in ["punkt", "punkt_tab", "stopwords", "wordnet", "omw-1.4"]:
    nltk.download(resource, quiet=True)

print("Project root:", ROOT)
print("Preprocessing tools are ready.")
```

**What you should see:** the project folder path, ending in `nlp-zero-to-hero`.

## Step 1 — Clean and tokenize a sentence


```python
from lessons.preprocessing import clean_text, tokenize, stem_tokens, lemmatize_tokens

sample = "I absolutely LOVE learning NLP! Visit https://example.com #nlp @user"

cleaned = clean_text(sample)
tokens = tokenize(cleaned, drop_stopwords=True)

print("Original:", sample)
print("Cleaned:", cleaned)
print("Tokens without stopwords:", tokens)
```

**What you should see:** the link, hashtag, and mention are removed. Very common words are removed too, so only the useful words remain.

## Step 2 — Compare stemming and lemmatization

Both methods reduce a word to a simpler form, but they do it differently:

- **Stemming** cuts the end of the word. It is fast but can produce a non-word.
- **Lemmatization** uses a dictionary and returns a real base word. It is slower but cleaner.


```python
words = ["studies", "studying", "better", "running"]

print("Stems:   ", stem_tokens(words))
print("Lemmas:  ", lemmatize_tokens(words))
```

**What you should see:** stems may look rough (`studi`), while lemmas stay closer to real words. The default lemmatizer treats words as nouns, so verb forms such as `studying` may not reduce unless you give it a part-of-speech hint.
